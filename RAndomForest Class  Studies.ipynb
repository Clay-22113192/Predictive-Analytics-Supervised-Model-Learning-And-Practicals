```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score,classification_report,r2_score
from sklearn.ensemble import RandomForestClassifier
```

```python
df=pd.read_csv('fruits.csv')
df.head()
```

```python
leb=LabelEncoder()
```

```python
df['shape']=leb.fit_transform(df['shape'])

df['color']=leb.fit_transform(df['color'])

df['taste']=leb.fit_transform(df['taste'])

df['fruit_name']=leb.fit_transform(df['fruit_name'])
```

```python
df.head(20)
```

```python
x=df.drop('fruit_name',axis=1)
y=df['fruit_name']
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
# random_state deermines the reproducibility of the splitting
```

```python
#creating the model. if you fail to bring the n_estimators, it is still going to be 100
model=RandomForestClassifier(random_state=42)
```

```python
model.fit(x_train,y_train)
```

```python
y_pred=model.predict(x_test)
y_pred
```

```python
model.score(x_test,y_test)
```

```python
accuracy=accuracy_score(y_test,y_pred)
print(accuracy)
```

```python
classificationReport=classification_report(y_test,y_pred)
print(classificationReport)
```
